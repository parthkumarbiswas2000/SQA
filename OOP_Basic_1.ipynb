## Methods/Function


```python
class student:
  ID = ""
  CGPA = ""

  def settingvalve(self,ID,CGPA):
      self.ID = ID
      self.CGPA = CGPA

  def gettingvalve(self):
      print(f"ID: {self.ID}, CGPA: {self.CGPA}")

nusrat =  student()
nusrat.settingvalve(103,3.78)
nusrat.gettingvalve()

niladri = student()
niladri.settingvalve(104,3.21)
niladri.gettingvalve()
```

## Constructor

```python
class student:

  def __init__(self,ID,cgpa):
    self.ID = ID
    self.cgpa = cgpa

  def show_info(self):
      print(f"ID: {self.ID}, CGPA: {self.cgpa}")

alice = student(105,3.6)
alex = student(106,3.9)

alice.show_info()
alex.show_info()
```

## Destructor

```python
class student:

  def __init__(self,Name,ID):
    self.Name= Name
    self.ID=ID
    print(f"Student object created {self.Name}, (ID:{self.ID})")

  def __del__(self):
    print(f"Student object destroyed {self.Name}, (ID:{self.ID})")


T1 = student("Farzana",101)
T2 = student("Sakib",102)


del T1

print("End of program......")
```

## Inheritance

```python
class phone:

  def call(self):
    print("You can call")

  def message(self):
    print("You can message")

class samsung:
  def call(self):
    print("You can call")
  def message(self):
    print("You can message")

  def message(self):
    print("You can message")

  def selfie(self):
    print("You can take selfie")

p = phone()
p.call()
p.message()

s = samsung()
s.call()
s.message()
s.selfie()
```

## Method Overriding

```python
class phone:
  def __init__(self):
    print("I'm in phone class")

class samsung(phone):
  def __init__(self):
    super().__init__()
    print("I'm in samsung class")

s = samsung()
```

## the area calculation of rectangle and triangle

```python
class shape:
  def __init__(self,dim1,dim2):
    self.dim1 = dim1
    self.dim2 = dim2

  def area(self):
    print("I'm Area method of shape")

class rectangle(shape):
  def area(self):
    print(f"Area of rectangle: {self.dim1*self.dim2}")

class triangle(shape):
  def area(self):
    print(f"Area of triangle: {0.5*self.dim1*self.dim2}")

r = rectangle (10,5)
r.area()

t = triangle (10,5)
t.area()
```

## Abstraction

```python
from abc import ABC, abstractmethod

class shape(ABC):
  def __init__(self,dim1,dim2):
    self.dim1 = dim1
    self.dim2 = dim2

  @abstractmethod
  def area(self):
    pass

class triangle(shape):
  def area(self):
    print(f"Area of triangle: {0.5*self.dim1*self.dim2}")

class rectangle (shape):
  def area(self):
    print(f"Area of rectangle: {self.dim1*self.dim2}")

t = triangle (10,5)
t.area()
r = rectangle (10,5)
r.area()
```
